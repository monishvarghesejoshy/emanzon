```python
input1=int(input("enter first digit"))

match input1:
    case 1:
        print("one")
    case 2:
        print("two")
    case 3:
        print("three")
    case 4:
        print("four")
    case 5:
        print("five")
    case _ :
        print("invalid")
```

```python
for i in range(2,30,1):
  for j in range(2,i):
     if (i%j) == 0:
      print(i, "is not a prime")
      break
  else:
    print(i, "is a prime")
```
