```python

def greater(input_list: list)->list:
    maxsalary=max(salary for name,salary in input_list)
    return [(name, salary) for name, salary in input_list if
            salary == maxsalary ]
```

```python
emp=[('dave',3500),('sam',5000)]
```

```python
greater(emp)
```
